```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import os
import cv2
from tensorflow.keras.layers import Dense, Flatten, Conv2D, Dropout, MaxPooling2D
from tensorflow.keras.models import Sequential
import warnings
warnings.filterwarnings('ignore')
```

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("cashutosh/gender-classification-dataset")

print("Path to dataset files:", path)
```

```python
path_train_male=os.path.join(path,'Training','male')
path_train_female=os.path.join(path,'Training','female')
path_val_male= os.path.join(path, 'Validation', 'male')
path_val_female = os.path.join(path, 'Validation', 'female')
```

```python
list_train_male=os.listdir(path_train_male)
training_male=[[cv2.imread(os.path.join(path_train_male,list_train_male[i])),1] for i in range(len(list_train_male))]
```

```python
list_train_female=os.listdir(path_train_female)
training_female=[[cv2.imread(os.path.join(path_train_female,list_train_female[i])),0] for i in range(len(list_train_female))]
```

```python
training=training_male+training_female
```

```python
list_val_male=os.listdir(path_val_male)
validation_male=[[cv2.imread(os.path.join(path_val_male,list_val_male[i])),1] for i in range(len(list_val_male))]
```

```python
list_val_female=os.listdir(path_val_female)
validation_female=[[cv2.imread(os.path.join(path_val_female,list_val_female[i])),0] for i in range(len(list_val_female))]
```

```python
validation=validation_male+validation_female
```

```python
import random
plt.figure(figsize=(16,8))
for i in range(4):
    plt.subplot(1,4,i+1)
    index=random.randint(0,len(training))
    c=training[index][1]
    plt.imshow(training[index][0])
    if c==1:
        plt.title("Male", color='red')
    else :
        plt.title("Female", color='red')
```

```python
for i in range(len(training)):
    image_resize=cv2.resize(training[i][0],(100,100))
    training[i][0]=image_resize
```
